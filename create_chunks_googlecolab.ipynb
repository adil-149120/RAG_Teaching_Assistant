```python
import whisper
import json
import os
from google.colab import drive
drive.mount("/content/drive")

model=whisper.load_model("large-v2")

audios = os.listdir("/content/drive/MyDrive/RAG-project/Audios")

for audio in audios:
    number=audio.split("_")[0]
    title=audio.split("_")[1][:-4]
    print(number,title)
    result=model.transcribe(audio=f"/content/drive/MyDrive/RAG-project/Audios/{audio}",
    #result=model.transcribe(audio="Audios/0_sample.mp3",
                            language="hi",
                            task="translate",
                            word_timestamps=False)

    chunks=[]
    for segment in result["segments"]:
        chunks.append({"number" : number ,"title" : title ,"start" : segment["start"],"end" : segment["end"],"text" : segment["text"]})

    chunks_with_metadata={"chunks" : chunks , "text" : result["text"]}

    output_path = f"/content/drive/MyDrive/RAG-project/jsons/{audio}.json"

    with open(output_path, "w", encoding="utf-8") as f:
        json.dump(chunks_with_metadata, f, ensure_ascii=False, indent=2)

    print("Saved:", output_path)

print("\n\nTask Done : SUCCESSFUL\n\n")

```
